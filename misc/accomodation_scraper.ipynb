```python
import os
import requests
import numpy as np
import pandas as pd
from bs4 import BeautifulSoup
import time
```

```python
def get_ir_content(idx):
    url  = f'http://ir.speakingsame.com/rp.php?q=Maynooth%2C+Co.+Kildare&id={idx}'#81033'
    page = requests.get(url)
    soup = BeautifulSoup(page.content, "html.parser")
    results = soup.find(id="mainT")
    res_txt = results.text.lower()
    return results.text.lower()
```

```python
tic = time.time()
rent_desc = dict()
for key in range(81030, 81040):#[81033, 8811818, 81034, 81036]:
    rent_desc[key] = get_ir_content(key)
toc = time.time()
print(toc - tic)
```
