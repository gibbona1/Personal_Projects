```python
fnames = df.file_name.unique()#[0]
fnames_pre = [fname[:fname.find('_start')]+'.wav' for fname in fnames]
fnames_pre
#fname
```

```python
all_files = []
for f in wf_folders:
    all_files.extend([os.path.join(f, 'Data2', fname) for fname in os.listdir(os.path.join(f, 'Data2'))])
```

```python
all_files
```

```python
#tuple(fnames_pre)
inlabs_files = [f for f in all_files if f.endswith(tuple(fnames_pre))]
inlabs_files
```

```python
sv_folder = 'wf_data_clips'
if not os.path.exists(sv_folder):
    os.mkdir(sv_folder)
```

```python
for fname, fname_pre in zip(fnames, fnames_pre):
    print(fname_pre)
    if fname.startswith(tuple([str(i)+'_' for i in range(1,10)])):
        continue
    file_name = [f for f in inlabs_files if f.endswith(fname_pre)][0]
    pre, ext  = os.path.splitext(file_name)
    samp_rate, audio = wavfile.read(file_name)
    file_dur = len(audio)#/samp_rate
    minutes = int(fname.split('_')[-2])
    #print(fname)
    #print(fname.split('_')[-1])
    #print(fname.split('_')[-1][:-4])
    seconds = int(fname.split('_')[-1][:-4])
    st_pos = int((minutes*60+seconds)*samp_rate)
    en_pos = int(st_pos + 15*samp_rate)
    audio_samp   = audio[st_pos:en_pos]
    sav_str = os.path.join(sv_folder, fname)
    wavfile.write(sav_str, samp_rate, audio_samp)
```
