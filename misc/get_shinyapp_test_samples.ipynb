```python
import math
def splitter_fn(startArray):
    secLength = math.ceil(len(startArray) / 3)
    newArray = []
    arr_a = startArray[:secLength]
    arr_a.sort()
    newArray.append(arr_a)
    arr_b = startArray[secLength : 2*secLength]
    arr_b.sort()
    newArray.append(arr_b)
    arr_c = startArray[2*secLength:]
    arr_c.sort()
    newArray.append(arr_c)
    return newArray
```
