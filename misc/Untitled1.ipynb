```python
prop_df = pd.read_csv('RTB_Register_Data.csv')
print("num records:"+str(len(prop_df)))
prop_df.head()
```

```python
address_cols = ['AddressLine'+str(l) for l in range(1,6)]
str_cols = address_cols+['Eircode', 'County']
int_cols = ['NoOfBedrooms']
```

```python
prop_df[str_cols] = prop_df[str_cols].fillna('')
prop_df[int_cols] = prop_df[int_cols].fillna(0)
```

```python
prop_df = prop_df[prop_df['NoOfBedrooms'] == 2]
print("num records:"+str(len(prop_df)))

prop_df.head()
```

```python
for ac in address_cols:
    prop_df = prop_df.loc[~prop_df[ac].str.contains("maynooth road", case=False)]
print("num records:"+str(len(prop_df)))
prop_df.head()
```

```python
prop_df['AddressFull'] = prop_df[address_cols].apply(", ".join, axis=1)
for _ in range(3):
    prop_df['AddressFull'] = prop_df['AddressFull'].str.replace(', ,', ',')
for _ in range(2):
    prop_df['AddressFull'] = prop_df['AddressFull'].str.rstrip(' ')
    prop_df['AddressFull'] = prop_df['AddressFull'].str.rstrip(',')
prop_df[['AddressFull', 'Eircode']].head()
```
