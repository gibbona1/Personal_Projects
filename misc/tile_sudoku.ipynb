```python
def possible(y,x,n):
    global grid
    # n is the number we want to fill in
    N = len(grid)
    # 1st 
    # check if n already existed in vertical (y) axis
    # if exists, return False (not possible)
    for i in range(N):
        if grid[y][i] == n:
            return False

    # 2nd
    # check horizontal (x) axis
    for i in range(N):
        if grid[i][x] == n:
            return False

    # 3rd
    # check if it points to anything
    if x > 0:
        if grid[y][x - 1] != 0:
            y0 = grid[y][x - 1] - 1
            x0 = n-1
            print(grid[y0][x0], y+1)
            if grid[y0][x0] not in [0, y+1]:
                return False
    elif x < N-1:
        if grid[y][x + 1] != 0:
            y0 = n-1
            x0 = grid[y][x + 1] - 1
            print(grid[y0][x0], y+1)
            if grid[y0][x0] not in [0, y + 1]:
                return False

    # return true if pass all 3 checks.
    return True
```

```python
def solve(iter = 0):
  global grid
  #if iter > 1000:
  #  print(f"exiting after {iter} iterations")
  #  return grid
  N = len(grid)
  for y in range(N):
    for x in range(N):
      #print(f'iter: {iter}, y: {y}, x: {x}, grid: {grid[y][x]}')
      # Find blank positions in the grid (value = 0)
      if grid[y][x] == 0:
        # Loop n from 1-8
        for n in range(1, N + 1):
          if possible(y, x, n):
            grid[y][x] = n
            solve(iter + 1)

            # This is where backtracking happens
            # Reset the latest position back to 0 and try with new n value
            grid[y][x] = 0
        return
  print(np.matrix(grid))
  input('More?')
```
